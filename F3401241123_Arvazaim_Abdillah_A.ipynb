Basic

```python
# ════════════════════════════════════════════════════════════
#  SOAL BASIC
# ════════════════════════════════════════════════════════════

# ── A. Buat class "Panen" ──────────────────────────────────
class Panen:

    # ── B. Definisikan atribut private ────────────────────
    def __init__(self, nama_kebun, luas_lahan, hasil_panen):
        self.__nama_kebun  = nama_kebun   # private
        self.__luas_lahan  = luas_lahan   # private
        self.__hasil_panen = hasil_panen  # private

    # ── C. Method publik get_luas() dan get_hasil() ───────
    def get_luas(self):
        return self.__luas_lahan

    def get_hasil(self):
        return self.__hasil_panen


# ── D. Definisikan objek berdasarkan data ─────────────────
kebun1 = Panen("Melati",  15, 7500)
kebun2 = Panen("Anggrek", 20, 9200)
kebun3 = Panen("Mawar",   12, 6100)
kebun4 = Panen("Kenanga", 18, 8500)

# Simpan seluruh objek ke dalam list bernama "kebun"
kebun = [kebun1, kebun2, kebun3, kebun4]

# Tampilkan informasi menggunakan looping
print("=" * 50)
print("         INFORMASI PANEN KEBUN TEH")
print("=" * 50)

for i, k in enumerate(kebun, start=1):
    # Nama kebun diakses sebagai atribut private melalui name-mangling
    print(f"Kebun ke-{i}")
    print(f"  Nama Kebun   : {k._Panen__nama_kebun}")   # atribut private
    print(f"  Luas Lahan   : {k.get_luas()} ha")         # method publik
    print(f"  Hasil Panen  : {k.get_hasil()} kg")        # method publik
    print("-" * 50)
```

Intermediate

```python
# ════════════════════════════════════════════════════════════
#  SOAL INTERMEDIATE
# ════════════════════════════════════════════════════════════

# ── A. Class Panen_Jual (turunan dari Panen) ───────────────
class Panen_Jual(Panen):

    def __init__(self, nama_kebun, luas_lahan, hasil_panen, harga=0):
        # Mewarisi __init__ dari class induk
        super().__init__(nama_kebun, luas_lahan, hasil_panen)
        self.__harga = harga   # atribut tambahan

    # ── B. Method get_harga() ─────────────────────────────
    def get_harga(self):
        return self.__harga

    # ── C. Polymorphism — override get_luas() dan get_hasil() ──
    def get_luas(self):
        luas = super().get_luas()
        print(f"  [INFO PENJUALAN] Luas lahan yang dijual : {luas} ha")
        return luas

    def get_hasil(self):
        hasil = super().get_hasil()
        print(f"  [INFO PENJUALAN] Hasil panen yang dijual: {hasil} kg")
        return hasil


# ── D. Buat objek menggunakan class Panen_Jual ────────────
panen_jual1 = Panen_Jual("Melati",  15, 7500, 18_000_000)
panen_jual2 = Panen_Jual("Anggrek", 20, 9200, 22_000_000)
panen_jual3 = Panen_Jual("Mawar",   12, 6100, 15_500_000)
panen_jual4 = Panen_Jual("Kenanga", 18, 8500, 20_000_000)

daftar_jual = [panen_jual1, panen_jual2, panen_jual3, panen_jual4]

# ── E. Panggil get_hasil() dan get_luas() untuk setiap objek ─
print("\n" + "=" * 50)
print("     INFORMASI PENJUALAN HASIL PANEN TEH")
print("=" * 50)

for i, pj in enumerate(daftar_jual, start=1):
    print(f"\nKebun ke-{i} — {pj._Panen__nama_kebun}")
    pj.get_hasil()
    pj.get_luas()
    print(f"  Harga Jual   : Rp {pj.get_harga():,.0f}")
    print("-" * 50)
```

Advance

```python
# ════════════════════════════════════════════════════════════
#  SOAL ADVANCE
# ════════════════════════════════════════════════════════════

import sqlite3
import pandas as pd

# ── A. Buat database SQL: PanenTeh.db ─────────────────────
conn   = sqlite3.connect("PanenTeh.db")
cursor = conn.cursor()

# ── B. Buat tabel KatalogPanen ────────────────────────────
cursor.execute("""
    CREATE TABLE IF NOT EXISTS KatalogPanen (
        KodePanen  TEXT PRIMARY KEY,
        NamaKebun  TEXT NOT NULL,
        LuasLahan  INTEGER NOT NULL,
        HasilPanen INTEGER NOT NULL,
        HargaJual  INTEGER NOT NULL,
        Stok       INTEGER NOT NULL
    )
""")
conn.commit()

# ── C. Masukkan seluruh data ──────────────────────────────
data_panen = [
    ("P001", "Melati",  15, 7500, 18_000_000, 4),
    ("P002", "Anggrek", 20, 9200, 22_000_000, 6),
    ("P003", "Mawar",   12, 6100, 15_500_000, 3),
    ("P004", "Kenanga", 18, 8500, 20_000_000, 5),
]

cursor.executemany("""
    INSERT OR IGNORE INTO KatalogPanen
    (KodePanen, NamaKebun, LuasLahan, HasilPanen, HargaJual, Stok)
    VALUES (?, ?, ?, ?, ?, ?)
""", data_panen)
conn.commit()

# ── D. Update harga kebun Anggrek menjadi Rp 25.000.000 ───
cursor.execute("""
    UPDATE KatalogPanen
    SET HargaJual = 25000000
    WHERE NamaKebun = 'Anggrek'
""")
conn.commit()

# ── E. Tampilkan tabel yang telah diperbarui via pandas ───
df = pd.read_sql_query("SELECT * FROM KatalogPanen", conn)

# Format kolom harga agar mudah dibaca
df["HargaJual"] = df["HargaJual"].apply(lambda x: f"Rp {x:,.0f}")

print("\n" + "=" * 60)
print("         DATABASE KATALOG PANEN TEH (Setelah Update)")
print("=" * 60)
print(df.to_string(index=False))

conn.close()
print("\n[✓] Database PanenTeh.db berhasil dibuat dan diperbarui.")
```

Kesimpulan:

Soal Basic (Enkapsulasi): Program diimplementasikan menggunakan kelas Panen yang menerapkan prinsip enkapsulasi. Atribut nama_kebun, luas_lahan, dan hasil_panen didefinisikan sebagai atribut private guna membatasi akses langsung dari luar, sehingga data hanya dapat diakses melalui metode publik get_luas() dan get_hasil(). Seluruh data kebun kemudian dikelola dalam bentuk list dan ditampilkan secara sistematis melalui perulangan (looping).

Soal Intermediate (Inheritance & Polymorphism): Konsep pewarisan (inheritance) diterapkan melalui kelas Panen_Jual sebagai turunan dari kelas Panen. Kelas ini mewarisi seluruh fungsi induk sekaligus menambahkan atribut harga dan metode get_harga(). Selain itu, penerapan polymorphism dilakukan dengan melakukan overriding pada metode get_luas() dan get_hasil(), yang memungkinkan penambahan informasi spesifik terkait penjualan tanpa mengubah struktur kelas aslinya.

Soal Advance (Database & Pandas): Tahap ini berfokus pada integrasi basis data menggunakan sqlite3 untuk mengelola persistensi data ke dalam file PanenTeh.db. Setelah tabel KatalogPanen berhasil dibentuk dan diisi, dilakukan manipulasi data berupa pembaruan (update) harga pada kebun "Anggrek". Hasil akhir data tersebut kemudian divisualisasikan secara rapi dan terstruktur dengan bantuan library pandas.
